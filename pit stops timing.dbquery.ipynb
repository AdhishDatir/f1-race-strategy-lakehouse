```sql
SELECT
    race_id,
    driver_id,
    pit_stop_count,
    ROUND(average_stop_duration_ms / 1000, 2)
        AS average_stop_duration_seconds,
    ROUND(fastest_stop_duration_ms / 1000, 2)
        AS fastest_stop_duration_seconds,
    first_pit_lap,
    last_pit_lap
FROM workspace.f1_gold.pit_stop_performance
ORDER BY average_stop_duration_seconds;
```
