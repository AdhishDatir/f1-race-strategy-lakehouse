```python
catalog = spark.sql("SELECT current_catalog() AS catalog").first()["catalog"]

schemas = [
    "f1_bronze",
    "f1_silver",
    "f1_gold"
]

for schema in schemas:
    spark.sql(
        f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}"
    )

print("Lakehouse schemas created successfully.")
```
